```python
import random
import time
import numpy as np
import matplotlib.pyplot as plt

# Function to simulate using Python loop
def simulate_loop(days):
    arrivals = []
    for _ in range(days):
        arrivals.append(random.randint(10, 50))
    return arrivals

# Function to simulate using NumPy
def simulate_numpy(days):
    return np.random.randint(10, 51, size=days)

# Days to simulate
days_list = [100, 1000, 10000, 100000]

# Record execution times
loop_times = []
numpy_times = []

for days in days_list:
    # Loop implementation
    start = time.perf_counter()
    simulate_loop(days)
    loop_times.append(time.perf_counter() - start)

    # NumPy implementation
    start = time.perf_counter()
    simulate_numpy(days)
    numpy_times.append(time.perf_counter() - start)

# Print execution times
print("Execution Times:")
print("Days\tLoop (s)\tNumPy (s)")
for i, days in enumerate(days_list):
    print(f"{days}\t{loop_times[i]:.4f}\t{numpy_times[i]:.4f}")

# Plot graph
plt.plot(days_list, loop_times, label='Loop')
plt.plot(days_list, numpy_times, label='NumPy')
plt.xlabel('Number of Days')
plt.ylabel('Execution Time (s)')
plt.title('Simulation Execution Time Comparison')
plt.legend()

# Save graph as PNG
plt.savefig('simulation_comparison.png', bbox_inches='tight')
plt.show()
```
